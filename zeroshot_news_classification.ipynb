```python
import pandas as pd
from transformers import pipeline
from tqdm import tqdm

tqdm.pandas()
```

```python
news_data = pd.read_csv('AAPL_NEWS_2024_2025.csv', index_col=0)
news_data
```

```python
classifier = pipeline('zero-shot-classification', model='facebook/bart-large-mnli')
```

```python
news_data.iloc[1, ]['title']
```

```python
candidate_labels = ['bullish', 'bearish']
classifier(news_data.iloc[1, ]['title'], candidate_labels)
```

```python
news_data.iloc[91, ]['title']
```

```python
candidate_labels = ['non-geopolitical', 'geopolitical']
classifier(news_data.iloc[91, ]['title'], candidate_labels)
```

```python
candidate_labels = ['prediction', 'reporting', 'other']
classifier(news_data.iloc[91, ]['title'], candidate_labels)
```

```python
sentiment_labels = ['positive', 'negative', 'neutral']
market_labels = ['bullish', 'bearish', 'neutral']
economic_labels = ['geopolitical', 'non-geopolitical']
outlook_labels = ['prediction', 'already happened', 'other']
```

```python
classifier(news_data.iloc[1, ]['title'], sentiment_labels)
```

```python
classifier(news_data.iloc[1, ]['title'], market_labels)
```

```python
classifier(news_data.iloc[1, ]['title'], economic_labels)
```

```python
classifier(news_data.iloc[2, ]['title'], outlook_labels)
```

```python
news_data_sample = news_data.head(20)
news_data_sample = news_data_sample[['publishedDate', 'title']]
news_data_sample
```

```python
news_data_sample['sentiment'] = news_data_sample['title'].progress_apply(lambda x: classifier(x, sentiment_labels)['labels'][0])
```

```python
news_data_sample
```

```python
news_data_sample['market_outlook'] = news_data_sample['title'].progress_apply(lambda x: classifier(x, market_labels)['labels'][0])
news_data_sample['economic'] = news_data_sample['title'].progress_apply(lambda x: classifier(x, economic_labels)['labels'][0])
news_data_sample['article_nature'] = news_data_sample['title'].progress_apply(lambda x: classifier(x, outlook_labels)['labels'][0])
```
